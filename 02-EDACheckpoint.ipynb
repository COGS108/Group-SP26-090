# Research Question

To what extent do undergraduate major categories, particularly fields related to Science and Engineering, predict stability in early-career employment among U.S. college graduates since 2016? This study evaluates how varying major fields in Science and Engineering differ in employment versus unemployment rates and occupation major match, assessing whether certain major categories are associated with stronger workforce tenure or greater mismatch vulnerabilities. A secondary goal would be assessing whether certain major categories are associated with stronger workforce tenure or greater mismatch vulnerabilities as an additional indicator of employment stability.

## Background and Prior Work

Job security is a common concern among college students and graduates. Our team shares the same concern, especially due to the increasingly competitive job market that makes it more difficult than ever for students to secure a position of employment. When discussing the changing state of the job market, one main factor that needs to be discussed is AI, which has drastically affected the job market by its ability to automate many tasks. Harvard Business Review’s research team categorized over 19,000 job tasks across over 900 positions to evaluate the potential for AI to automate the task. Through the interactive version of the data, it can be noticed that higher-level jobs that college graduates are more likely to pursue (one example being logistics analysts) have a higher calculated probability of automation than physical labor (an example being cashiers).<a name="cite_ref-1"></a>[<sup>[1]</sup>](#cite_note-1) However, it should also be noted that the authors found many positions that would be augmented, or improved, by generative AI. With these contrasting details, it becomes difficult to pinpoint how exactly AI changes the job market.

Though it is difficult to pinpoint exactly how AI affects the job market, what can be inferred is that the job market has gotten more competitive because of AI. As we have seen AI be capable of automating certain tasks and augmenting other positions, employers are expecting employees with more refined skills. On a study that NACE (National Association of Colleges + Employers) conducted between students that participated in experiential learning opportunities like internships versus those who did not, they found that 78.8% of experiential learners have a network in their early career compared to 64.3% of nonexperiential learners. <a name="cite_ref-2"></a>[<sup>[2]</sup>](#cite_note-2) The study highlights some of the differences between students already present in the more modern job market, where it becomes more crucial to have prior experience and knowledge to ensure job security (along with other differing factors like job satisfaction). # Hypothesis


We predict that undergraduate major categories, particularly fields related to Science and Engineering, will significantly relate to early-career employment stability among U.S. college graduates since 2016. Specifically, if we were to compare graduates from non-STEM fields to STEM fields, then we can expect non-STEM fields to exhibit higher unemployment rates and greater occupational major mismatch compared to graduates in STEM, engineering, and health-related fields. - N2OCMLS: occupation cod for last job (major occupational group)
- OCEDRLP: degree relatedness to job
- SALARY: annualized salary before deductions

Despite its strengths, the dataset has some limitations. It is cross-sectional rather than longitudinal, making it harder to track individual employment trajectories over time. Additionally, many key measures rely on self-reported responses, which may introduce subjective bias. Occupational and major categories are broad, which can obscure variation within fields. Salary data is also self-reported.

```python
# (D): Missing Data
# convert 'object' columns to floats to differentiate between Logical skip or actual Nan. 
numeric_cols = ['AGE', 'AGEGR', 'BAACYR', 'LFSTAT', 'SALARY', 'N2OCMLST', 'N2OCPRMG', 'NBAMEMG']
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors = 'coerce')
    
# convert different codes to Nan for specific columns for consistency
# unmask missing codes and logical skips. 
specific_replacements = {
    'BAACYR' : [9998, 9999],
    'N2OCMLST' : [8],
    'N2OCPRMG' : [8],
    'NBAMEMG' : [8],
    'SALARY' : [9999998, 9999999]
}

for col, codes in specific_replacements.items():
    if col in df.columns:
        df[col] = df[col].replace(codes, np.nan)

# replaces 'L' logical skips in all columns
df = df.replace('L', np.nan)

# PROOF of Systematic Missingness:
# missingness in salary is tied to LFSTAT 3 (not in the labor force). 
missing_sal_mask = df['SALARY'].isnull()
print("Evidence of Systematic Missingness: Labor Force Status for Missing Salaries: ")
print(df[missing_sal_mask]['LFSTAT'].value_counts())

print("\nOccupation Code (N2OCMLST) for missing salaries: ")
print(df[missing_sal_mask]['N2OCMLST'].value_counts(dropna=False))

# around 13K people with missing salaries are coded as 3 in LFSTAT
# 787 Nan means double missing responses. Never worked so no last occupation to report and no salary to report.
# verification that 787 people never worked . 
double_missing = df[df['SALARY'].isnull() * df['N2OCMLST'].isnull()]
print(f"Number of 'Double Missing' respondends: {len(double_missing)}")
# justification to drop
```

```python
# (E): Statistical Outliers
# flag those who are economic outliers (99th percentile cap on extreme high-earners)
sal_cap = df['SALARY'].quantile(0.99)
df['EXTREME_SALARY_FLAG'] = df['SALARY'] > sal_cap

# flag those who graduate before 2016, our year of measurement
df['OLD_GRAD_FLAG'] = df['BAACYR'] < 2016

# flag those who are not in labor force
df['NOT_IN_LABOR_FORCE_FLAG'] = df['LFSTAT'] == 3

# flag those outside target age range (19-30)
df['TARGET_AGE_FLAG'] = (df['AGE'] < 19) | (df['AGE'] > 30)
```

```python
#(F): Cleanup Data:
# chose to filter age cohort (19-30), graduate 2016+
# and drop those who are not in the labor force or missing salary data.

df_processed = df[
    (df['AGE'] >= 19) & (df['AGE'] <= 30) &
    (df['BAACYR'] >= 2016) &
    (df['LFSTAT'] != 3) &
    (df['SALARY'].notnull()) &
    (df['SALARY'] > 0)
].copy()

#Justification:
# The dataset was cleaned to focus on Early Career cohorts
# Ensure that analysis would reflect a more modern job market for new grads.
# Those not in the work force do not provide relevant data for employment study.
```

#### Bachelor's Major vs. Early-Career Salary

We investigated whether undergraduate major category predicts early career annual salary among US graduates aged 19-30 (class of 2016+, n = 11206). Since salary distributions were heavily right skewed across most major groups (ranging from 0.73 to 9.18), we use the Kruskal-Wallis test instead of ANOVA because it's heavily skewed. A Tukey post hoc test is then applied to identify which specific pairs of majors differ significantly. The box plot shows the full salary distribution per major ordered by median, while the chart shows the mean ± 95% confidence interval with sample size annotated.

The Kruskal-Wallis test confirms salary distributions differ significantly across major groups (H = 2023.19, p < 0.001). From the Tukey post hoc results Computer & Math (median of 92k) and Engineering (median of 91k) are each significantly higher than all other groups and are also significantly different from each other. The remaining 5 groups, S&E Related, Physical Sci, Social Sci, Bio/Ag & Life Sci, and Non-S&E, cluster around 55-62k median range and are not significantly different from one another. This suggests that major category predicts salary primarily at the top end, with CS and Engineering standing out, while the majority of other fields show comparable early carrer earnings regardless if the major is STEM or non-STEM.

```python
## YOUR CODE HERE
## FEEL FREE TO ADD MULTIPLE CELLS PER SECTION
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# ── Load processed data ──────────────────────────────────────────────────────
df = pd.read_csv('data/01-interim/epcg23_interim_cleaned.csv', low_memory=False)

# ── Map NBAMEMG codes to readable labels ─────────────────────────────────────
major_map = {
    1: 'Computer &\nMath',
    2: 'Bio/Ag &\nLife Sci',
    3: 'Physical\nSci',
    4: 'Social Sci',
    5: 'Engineering',
    6: 'S&E Related',
    7: 'Non-S&E',
}

df['MAJOR_LABEL'] = df['NBAMEMG'].map(major_map)
df_eda = df[df['MAJOR_LABEL'].notna() & df['SALARY'].notna()].copy()

# ── 1. Check skewness to choose ANOVA vs Kruskal-Wallis ──────────────────────
skew_vals = df_eda.groupby('MAJOR_LABEL')['SALARY'].apply(lambda x: x.skew())
print("Skewness per major group:")
print(skew_vals.round(2))

# ── 2. Kruskal-Wallis test (robust to skew) ───────────────────────────────────
groups = [g['SALARY'].values for _, g in df_eda.groupby('MAJOR_LABEL')]
h_stat, p_val = stats.kruskal(*groups)
print(f"\nKruskal-Wallis: H = {h_stat:.2f}, p = {p_val:.4e}")

# ── 3. Tukey HSD post-hoc ────────────────────────────────────────────────────
tukey = pairwise_tukeyhsd(
    endog=df_eda['SALARY'],
    groups=df_eda['MAJOR_LABEL'],
    alpha=0.05
)
print("\nTukey HSD Post-Hoc Results:")
print(tukey.summary())

# ── 4. Box plot + mean ± 95% CI bar chart ────────────────────────────────────
order = (df_eda.groupby('MAJOR_LABEL')['SALARY']
         .median()
         .sort_values(ascending=False)
         .index.tolist())

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle("Bachelor's Major vs. Early-Career Salary (NSCG 2023)",
             fontsize=14, fontweight='bold', y=1.01)

# -- Box plot --
data_ordered = [df_eda[df_eda['MAJOR_LABEL'] == m]['SALARY'].dropna() for m in order]
bp = axes[0].boxplot(data_ordered, patch_artist=True, notch=True,
                     medianprops=dict(color='black', linewidth=2),
                     flierprops=dict(marker='o', markersize=2, alpha=0.3))
colors = plt.cm.Set2(np.linspace(0, 1, len(order)))
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
axes[0].set_xticklabels(order, rotation=30, ha='right', fontsize=9)
axes[0].set_ylabel('Annual Salary ($)', fontsize=11)
axes[0].set_title('Distribution by Major (Box Plot)', fontsize=11)
axes[0].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
axes[0].grid(axis='y', alpha=0.4)

# -- Bar chart with error bars (mean ± 95% CI) --
summary = (df_eda.groupby('MAJOR_LABEL')['SALARY']
           .agg(['mean', 'sem', 'count'])
           .loc[order])
summary['ci95'] = 1.96 * summary['sem']

axes[1].bar(range(len(order)), summary['mean'],
            yerr=summary['ci95'], capsize=5,
            color=colors, edgecolor='white', linewidth=0.8,
            error_kw=dict(ecolor='black', elinewidth=1.2))
axes[1].set_xticks(range(len(order)))
axes[1].set_xticklabels(order, rotation=30, ha='right', fontsize=9)
axes[1].set_ylabel('Mean Annual Salary ($)', fontsize=11)
axes[1].set_title('Mean Salary ± 95% CI by Major', fontsize=11)
axes[1].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
axes[1].grid(axis='y', alpha=0.4)

for i, (_, row) in enumerate(summary.iterrows()):
    axes[1].text(i, row['mean'] + row['ci95'] + 1500,
                 f"n={int(row['count'])}", ha='center', va='bottom', fontsize=7.5)

stat_note = f"Kruskal-Wallis: H = {h_stat:.2f}, p {'< 0.001' if p_val < 0.001 else f'= {p_val:.4f}'}"
fig.text(0.5, -0.03, stat_note, ha='center', fontsize=10, style='italic')

plt.tight_layout()
plt.show()

# ── 5. Tukey HSD Heatmap ─────────────────────────────────────────────────────
groups_u = sorted(df_eda['MAJOR_LABEL'].unique())
n = len(groups_u)
idx = {g: i for i, g in enumerate(groups_u)}
reject_matrix = np.zeros((n, n))
for row in tukey.summary().data[1:]:
    g1, g2 = row[0], row[1]
    reject_matrix[idx[g1], idx[g2]] = 1 if row[6] == True else 0
    reject_matrix[idx[g2], idx[g1]] = 1 if row[6] == True else 0

fig, ax = plt.subplots(figsize=(8, 6.5))
ax.imshow(reject_matrix, cmap='RdYlGn', vmin=0, vmax=1, aspect='auto')

for i in range(n):
    for j in range(n):
        if i == j:
            ax.text(j, i, '—', ha='center', va='center', fontsize=11, color='gray')
        else:
            sig = reject_matrix[i, j]
            txt = 'Sig.*' if sig else 'n.s.'
            col = 'white' if sig else '#555'
            ax.text(j, i, txt, ha='center', va='center', fontsize=8.5,
                    fontweight='bold' if sig else 'normal', color=col)

ax.set_xticks(range(n))
ax.set_yticks(range(n))
ax.set_xticklabels(groups_u, fontsize=8.5, rotation=30, ha='right')
ax.set_yticklabels(groups_u, fontsize=8.5)
ax.set_title("Tukey HSD Post-Hoc: Pairwise Salary Comparisons\n(Green = significant difference, α = 0.05)",
             fontsize=11, fontweight='bold', pad=12)
plt.tight_layout()
plt.show()
```

#### Bachelor's Major Category vs Labor Force Status



The stacked bar plot shows noticeable differences in labor force status across bachelor major categories among U.S. college graduates. Engineering and Computer/Mathematics majors displayed some of the highest proportions of employed respondents, while Non-S&E and S&E-related majors showed comparatively larger proportions of individuals not participating in the labor force. Across all categories, unemployment represented a relatively small proportion of respondents, although slight differences between fields were still visible. These patterns suggest that undergraduate field of study may relate to differences in early-career employment outcomes and workforce participation.

To further evaluate this relationship, a chi-square test of independence was conducted between bachelor major category and labor force status. The test produced a chi-square statistic of 371.64 with 12 degrees of freedom and a p-value of 3.7804e-72. The results indicate that labor force status distributions vary across major categories rather than appearing evenly distributed across all fields of study.

Overall, the findings align with the project hypothesis that undergraduate major categories are associated with differences in employment stability and workforce outcomes. In particular, majors with stronger technical or specialized career pathways, such as Engineering and Computer/Mathematics, appeared to have higher proportions of employed graduates, while broader Non-S&E categories showed somewhat larger proportions of respondents outside the labor force.




```python
## YOUR CODE HERE
## FEEL FREE TO ADD MULTIPLE CELLS PER SECTION
#checking data
#Bachelor major vs. labor force status  (categorical vs categorical) → clustered/stacked bar plot

print(df['NBAMEMG'].value_counts(dropna=False))
print(df['LFSTAT'].value_counts(dropna=False))

import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency

# remove rows with missing values
clean_df = df[df['NBAMEMG'].notna() & df['LFSTAT'].notna()].copy()

# rename categories for better reading 
major_labels = {
    1.0: 'Comp/Math',
    2.0: 'Bio Sciences',
    3.0: 'Physical Sci',
    4.0: 'Social Sci',
    5.0: 'Engineering',
    6.0: 'S&E Related',
    7.0: 'Non-S&E'
}

labor_labels = {
    1: 'Employed',
    2: 'Unemployed',
    3: 'Not in Labor Force'
}

#cleaning columns 
clean_df['Major Category'] = clean_df['NBAMEMG'].map(major_labels)
clean_df['Labor Status'] = clean_df['LFSTAT'].map(labor_labels)

# contingency table
table = pd.crosstab(
    clean_df['Major Category'],
    clean_df['Labor Status']
)

# chi-square test
chi2, p, dof, expected = chi2_contingency(table)

#counts to percentages for plot 
table_pct = table.div(table.sum(axis=1), axis=0) * 100

# reorder columns for cleaner look
table_pct = table_pct[
    ['Employed', 'Unemployed', 'Not in Labor Force']
]

# reorder majors 
major_order = [
    'Comp/Math',
    'Engineering',
    'Physical Sci',
    'Bio Sciences',
    'S&E Related',
    'Social Sci',
    'Non-S&E'
]

table_pct = table_pct.reindex(major_order)

# stacked bar plot 
ax = table_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

# formatting for plot 
plt.title('Labor Force Status by Bachelor Major Category', fontsize=14)
plt.xlabel('Field of Study', fontsize=12)
plt.ylabel('Percentage of Respondents', fontsize=12)

plt.xticks(rotation=15)

plt.legend(
    title='Labor Status',
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()

# statistical results
print(f"Chi-square statistic: {chi2:.2f}")
print(f"P-value: {p:.4e}")
print(f"Degrees of freedom: {dof}")
```
