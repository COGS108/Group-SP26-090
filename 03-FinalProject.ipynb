### Bachelor's Major vs. Job Occupation Category

To evaluate the relationship between undergraduate bachelor's degree and career, we conducted a Chi-Square Test of Independence between major category (NBAMEMG) and occupational group (N2OCPRMG). This test compares "Observed" data (job occupations) to "Expected" data (major categories).
- The Chi-Square Statistic test yielded x^2 = 15522.47. From a dataset of 11206 responses, this high value indicates a great distance from randomness. In other words, the occupation match of graduates into specific jobs is too consistent to be a coincidence.
- The P-value yielded p ‹ 0.001. We can reject the null hypothesis that these results happened by chance.

These two results confirmed that a relationship exists but not how strong the relationship is. To measure the strength of this association beyond the influence of a large sample size, we performed a Cramer's V test which resulted in a value of 0.4805. This indicates a strong association and that undergraduate major is related to workforce entry.

We chose to visualize these findings using a normalized heatmap. The heatmap shows that Computer Science/Math (65.0%) and Engineering (59.5%) have the highest retention, supporting our hypothesis that specialized technical STEM fields offer greater job pathway stability. On the other hand, 68.5% of Social Science major graduates transitioned into Non-S&E occupations. This disparity identifies a significant occupational mismatch in the early-career labor market. It also confirms that the stability of degree-to-career match is significantly lower for non-engineering and computer science/math fields.

```python
## Load EDA1 DF + Filter 
df_eda1 = df.copy()

label_map = {
    1: 'Computer/Math Science',
    2: 'Bio/Agricultural/Life Science',
    3: 'Physical & Related Science',
    4: 'Social & Related Science',
    5: 'Engineering',
    6: 'Science & Engineering Related',
    7: 'Non-Science & Engineering'
}

# Filter out Logical Skip and apply
df_eda1 = df_eda1[df_eda1['NBAMEMG'] < 8]
df_eda1 = df_eda1[df_eda1['N2OCPRMG'] < 8]
df_eda1['Major_Category'] = df_eda1['NBAMEMG'].map(label_map)
df_eda1['Job_Category'] = df_eda1['N2OCPRMG'].map(label_map)
```

```python
## Make Contingency Table + Chi-Square Test
tb = pd.crosstab(
    df_eda1['Major_Category'],
    df_eda1['Job_Category']
)

chi2, p, dof, expected = chi2_contingency(tb)

print(f"Chi-Square Stat: {chi2: .2f}")
print(f"P-value: {p: .4e}")
print("p-value is so small because chi-square is sensitive to sample size.\n")
print("chi-square stat shows that relationship exits but not its strenght.\n")
```

```python
## calculating Cramer's V to show strength of association
## Cramer's V (effect size) which is not sensitive to sample size
n = tb.sum().sum()
min_dim = min(tb.shape) - 1
cramers_v = np.sqrt(chi2 / (n * min_dim))
print(f"Cramer's V: {cramers_v: .4f}")
```

### Bachelor's Major vs. Early-Career Salary

We investigated whether undergraduate major category predicts early career annual salary among US graduates aged 19-30 (class of 2016+, n = 11206). Since salary distributions were heavily right skewed across most major groups (ranging from 0.73 to 9.18), we use the Kruskal-Wallis test instead of ANOVA because it's heavily skewed. A Tukey post hoc test is then applied to identify which specific pairs of majors differ significantly. The box plot shows the full salary distribution per major ordered by median, while the chart shows the mean ± 95% confidence interval with sample size annotated.

The Kruskal-Wallis test confirms salary distributions differ significantly across major groups (H = 2023.19, p < 0.001). From the Tukey post hoc results Computer & Math (median of 92k) and Engineering (median of 91k) are each significantly higher than all other groups and are also significantly different from each other. The remaining 5 groups, S&E Related, Physical Sci, Social Sci, Bio/Ag & Life Sci, and Non-S&E, cluster around 55-62k median range and are not significantly different from one another. From the histogram, we can se that that major category predicts salary primarily at the top end, with CS and Engineering standing out, while the majority of other fields show comparable early carrer earnings regardless if the major is STEM or non-STEM.

In contrast, the remaining five major categories form a much tighter cluster. Their box plots show substantial overlap in both interquartile ranges and overall distributions, indicating similar earning outcomes. This pattern is reinforced by the Tukey comparison matrix, where most pairwise comparisons among these five groups are not statistically significant. Notably, majors traditionally considered STEM, such as Physical Science and Bio/Life Science, do not earn significantly more than several non-STEM categories during the early-career period examined here.

The stacked bar plot shows noticeable differences in labor force status across bachelor major categories among U.S. college graduates. ### EDA4: Impact of Bachelor's Degree Field on Employment Stability

In this section, we explore the relationship between a graduate's field of study and their likelihood of changing jobs or employers. We are analyzing two categorical variables: 
1. Bachelor's Degree Field of Study (NBAMEMG)
2. Employment Status (EMSMI), which tracks whether the individual is with the same employer and/or in the same job compared to the previous reference period.

To analyze these variables, we will use two approachs:

1. Visual Analysis (100% Stacked Bar Chart): This allows us to easily compare the conditional probability of employment outcomes across different majors, standardizing the distinct sample sizes of each major field.
2. Statistical Testing (Chi-Square Test of Independence): While visual trends are helpful, this non-parametric test will determine if any observed variations in employment stability across different degree fields are statistically significant.

From the stacked bar chart, we saw that Computer Science / Math majors held the highest retention (same employer + same job). Nevertheless, the practical differences between the bars linking a graduate's major to their employment status were minor. The dominant trend across all majors is strong employer retention. Furthermore, job changes that occur within the same company likely reflect internal promotions rather than career instability.

```python
## Load EDA4 + Map
df_eda4 = df.copy()
df_eda4['NBAMEMG'] = pd.to_numeric(df_eda4['NBAMEMG'], errors='coerce')
df_eda4['EMSMI'] = pd.to_numeric(df_eda4['EMSMI'], errors='coerce')
df_eda4 = df_eda4[df_eda4['NBAMEMG'].notna() & df_eda4['EMSMI'].notna()].copy()
df_eda4['NBAMEMG'] = df_eda4['NBAMEMG'].astype(int)
df_eda4['EMSMI'] = df_eda4['EMSMI'].astype(int)

## Reuse label_map from EDA1 for major
employment_dict = {
    1: 'Same employer & same job', 
    2: 'Same employer BUT diff job',
    3: 'Diff employer BUT same job', 
    4: 'Diff employer & diff job'
}

df_eda4['Major_Category'] = df_eda4['NBAMEMG'].map(label_map)
df_eda4['Employment_Status'] = df_eda4['EMSMI'].map(employment_dict)
df_eda4 = df_eda4[df_eda4['Major_Category'].notna() & df_eda4['Employment_Status'].notna()].copy()
```

```python
# STATISTICAL TESTING: Chi-Square
# Frequency contingency table with raw counts
contingency_table = pd.crosstab(
    df_eda4['NBAMEMG'], 
    df_eda4['EMSMI']
)
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("\n--- Chi-Square Test Results ---")
print(f"Chi-Square Statistic: {chi2:.2f}")
print(f"P-value: {p}")
print("Conclusion: Statistically significant relationship exists (p < 0.05).")
```

## Discussion and Conclusion

Overall, our analyses revealed several important relationships between undergraduate major and early career outcomes. We found that major category was associated with occupation category, salary, employment status, and employment stability, which suggests that graduates in distinct  fields experience different career trajectories after graduation. These findings provide evidence that a student’s field of study is related to multiple aspects of early career success and employment outcomes. 

This project builds on previous research by examining career outcomes using a nationally representative sample of recent college graduates across the United States. While prior studies often focused on specific locations or individual measures of career success, our analysis considered multiple outcomes, including salary, employment status, employment stability, and the relationship between a graduate's major and occupation. By analyzing these factors together, our study provides a broader perspective on how undergraduate major is associated with early-career opportunities and outcomes in today's labor market.

Overall, our findings suggest that undergraduate major is associated with several important early-career outcomes, but the strength of that relationship varies depending on the outcome being examined. Engineering and Computer/Mathematics majors showed the strongest alignment between degree and occupation, the highest salaries, and some of the highest employment rates, indicating that these fields may provide clearer pathways into the workforce. In contrast, graduates from other majors were more likely to work outside of their field of study and generally earned lower salaries early in their careers.

However, at the same time, our employment stability analysis suggests that major may play a smaller role once graduates have entered the workforce. Although statistically significant differences existed across majors, employer retention was high across all fields, and many job changes may reflect career advancement rather than instability. Overall, we learned that undergraduate major appears to have the greatest impact on workforce entry and early-career opportunities, while longer-term employment stability may be influenced by a broader set of factors beyond field of study alone.

Overall, our findings largely aligned with our initial hypothesis: undergraduate major categories, particularly in fields related to Science and Engineering, demonstrated the highest levels of job security and degree-occupation match. However, we did encounter a nuanced, somewhat unexpected result regarding employment stability. While we hypothesized that non-STEM fields would exhibit significantly higher turnover or instability compared to specialized STEM fields, our analysis revealed that employer retention was relatively high across all fields once graduates entered the workforce.

This apparent contradiction could be a real-world example of Simpson's Paradox, driven by the inherent structure of our dataset. We worked with the most comprehensive data available, but the NSF intentionally aggregates all non-STEM majors into a single, broad category because the survey's primary focus is tracking the Science and Engineering (S&E) workforce. Because of this structural limitation, the overarching retention rate for the non-STEM category may be artificially inflated by heavily populated, highly stable non-STEM fields, such as education, business, or nursing. This aggregation can easily mask severe instability or high turnover in smaller fields like the arts or humanities. By looking at the NSF's combined data, the expected trend of non-STEM instability disappears, whereas if the variables had been separated into their distinct majors, it might have revealed the exact disparities we originally hypothesized. Additionally, job changes in non-STEM fields could reflect upward mobility and career advancement rather than forced instability or unemployment.

Despite the robust insights provided by the 2023 National Survey of College Graduates (NSCG), our work has several notable limitations:
- Grouping Bias and Granularity: Because the dataset’s primary focus is on the Science and Engineering workforce, all non-STEM fields were lumped into a single, broad category. This limits our ability to make generalized claims about non-STEM majors and obscures the nuanced realities within specific humanities, arts, or social science disciplines.
- Cross-Sectional vs. Longitudinal Data: Our dataset is cross-sectional, capturing a single snapshot of recent graduates. While this highlights current early-career outcomes, longitudinal data would be incredibly helpful to track individual employment trajectories over time and determine if initial job-major mismatches eventually resolve as careers progress.
- Temporal Changes and the AI Boom: There is strong evidence that the phenomena we are investigating changes rapidly over time. The job market is currently experiencing a massive shift due to artificial intelligence automating and augmenting various roles. Because our data is capped at 2023, it may not fully capture the most recent impacts of AI on entry-level hiring, meaning the employment landscape for the upcoming graduating cohort could look different than the trends modeled here.
- Self-Reporting Bias: Finally, key measures such as salary and employment status rely on self-reported survey responses, which inherently introduces the risk of subjective bias or reporting inaccuracies.

If this investigation were to continue outside the scope of this class, there are several compelling directions for future research:
- Longitudinal Tracking of Macro Events: A natural next step would be analyzing how major societal events (e.g., the 2008 recession, the COVID-19 pandemic, and the current AI boom) uniquely impact specific majors. We could investigate whether STEM graduates are more insulated from economic shocks compared to their non-STEM peers, or if tech-focused majors are now facing unprecedented instability due to recent industry layoffs.
- Granular Non-STEM Analysis: To build a more inclusive narrative, future work must acquire datasets that disaggregate non-STEM fields. Investigating why career outcomes for non-STEM majors are historically harder to track could reveal deeper insights into the non-linear career pathways these graduates pursue.
- Alternative Educational Pathways: Finally, we could expand our scope beyond bachelor's degrees. With growing advocacy for the skilled trades as a safeguard against white-collar AI automation, comparing the long-term job security and salary trajectories of trade school graduates against traditional bachelor’s degree holders would provide incredibly valuable insights for incoming students deciding on their educational future.
